```python
import pandas as pd
import numpy as np

# Make the results reproducible
np.random.seed(42)

# Number of transactions
n = 3000

# -----------------------------
# Basic transaction information
# -----------------------------

transaction_ids = [f"TXN{i:04d}" for i in range(1, n + 1)]

customer_ids = [
    f"CUST{np.random.randint(1000, 2000)}"
    for _ in range(n)
]

dates = pd.date_range(
    start="2026-01-01",
    end="2026-06-30"
)

transaction_dates = np.random.choice(dates, size=n)

amounts = np.round(
    np.random.lognormal(mean=7, sigma=0.7, size=n),
    2
)

# Keep transaction amounts between ₹100 and ₹20,000
amounts = np.clip(amounts, 100, 20000)

# -----------------------------
# Payment details
# -----------------------------

payment_methods = np.random.choice(
    ["UPI", "Credit Card", "Debit Card", "Net Banking", "Wallet"],
    size=n,
    p=[0.40, 0.22, 0.18, 0.12, 0.08]
)

devices = np.random.choice(
    ["Mobile", "Desktop", "Tablet"],
    size=n,
    p=[0.65, 0.28, 0.07]
)

locations = np.random.choice(
    ["Pune", "Mumbai", "Bangalore", "Delhi", "Hyderabad", "Chennai"],
    size=n
)

# -----------------------------
# Payment status
# -----------------------------

statuses = np.random.choice(
    ["Success", "Failed"],
    size=n,
    p=[0.88, 0.12]
)

# -----------------------------
# Failure reasons
# -----------------------------

failure_reasons = []

for status in statuses:
    if status == "Success":
        failure_reasons.append("None")
    else:
        failure_reasons.append(
            np.random.choice(
                [
                    "Network Error",
                    "Insufficient Funds",
                    "Bank Declined",
                    "Technical Error",
                    "Invalid Details"
                ],
                p=[0.30, 0.22, 0.20, 0.18, 0.10]
            )
        )

# -----------------------------
# Create DataFrame
# -----------------------------

df = pd.DataFrame({
    "Transaction_ID": transaction_ids,
    "Customer_ID": customer_ids,
    "Transaction_Date": transaction_dates,
    "Amount": amounts,
    "Payment_Method": payment_methods,
    "Device": devices,
    "Location": locations,
    "Status": statuses,
    "Failure_Reason": failure_reasons
})

# Sort by date
df = df.sort_values("Transaction_Date").reset_index(drop=True)

# Display first 10 rows
df.head(10)
```

```python
df.to_csv("payment_transactions.csv", index=False)

print("Dataset saved successfully!")
```
